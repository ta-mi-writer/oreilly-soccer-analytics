```python
# 基本的なインポートとバージョン確認
import sys
import pandas as pd
import numpy as np

print(f"Pythonバージョン: {sys.version}")
print(f"Pandasバージョン: {pd.__version__}")
print(f"NumPyバージョン: {np.__version__}")
```

## データの作成と確認

```python
# 簡単なDataFrameを作成して確認
df = pd.DataFrame({
    "name": ["Alice", "Bob", "Charlie", "Diana"],
    "age": [25, 30, 35, 28],
    "score": [90.5, 85.0, 92.3, 88.7]
})

print("DataFrameの内容:")
df
```

```python
# 基本的な統計量の確認
print("基本統計量:")
df.describe()

# 年齢の平均値
print(f"\n平均年齢: {df['age'].mean():.1f}")
# スコアの最高値
print(f"最高スコア: {df['score'].max():.1f}")
```

## 簡単なテスト

```python
# 簡単なアサーションテスト
assert df.shape == (4, 3), "DataFrameの形状が正しくありません"
assert df['age'].mean() > 0, "年齢の平均が0以下です"
assert df["score"].max() <= 100, "スコアが100を超えています"

print("✅ すべてのテストに合格しました!")
```
