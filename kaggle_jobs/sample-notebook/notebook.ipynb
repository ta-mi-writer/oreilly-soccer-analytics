# サンプル・マルチセル・Kaggleノートブック

このノートブックは、複数のセルに処理を分割し、前のセルの結果（変数やファイル）を次のセルで引き継いで利用するシンプルなサンプルです。

### セル構成
1. **環境確認 & 設定変数定義**: ワーキングディレクトリの確認と、処理で使用する共通変数を定義します。
2. **データ生成 (セル1の出力)**: ダミーのデータやファイルを作成し、次のセルに渡します。
3. **データ処理 (セル2の処理)**: 1つ目のセルで作成されたファイルや変数を読み込んで集計・処理を行います。

```python
# ========================================
# 1. 環境確認 & 設定変数定義
# ========================================
from pathlib import Path
import os

# Kaggle環境でのワーキングディレクトリ確認
WORKING_DIR = Path("/kaggle/working") if os.path.exists("/kaggle/working") else Path("./output")
WORKING_DIR.mkdir(parents=True, exist_ok=True)

OUTPUT_FILE = WORKING_DIR / "sample_data.txt"
MESSAGE = "Kaggle Multi-Cell Pipeline Test Successful!"

print(f"Working Directory: {WORKING_DIR}")
print(f"Target Output File: {OUTPUT_FILE}")
```

```python
# ========================================
# 2. データ生成 (セル1の変数・パスを使用)
# ========================================
# 1つ目のセルで定義された `OUTPUT_FILE` と `MESSAGE` を使ってファイルを出力します

print(f"Writing message to {OUTPUT_FILE}...")
OUTPUT_FILE.write_text(MESSAGE, encoding="utf-8")

print("Data generated successfully!")
```

```python
# ========================================
# 3. データ処理 (セル2の結果を読み込んで処理)
# ========================================
# 2つ目のセルで生成されたファイルを確認し、内容を読み取って処理します

if OUTPUT_FILE.exists():
    content = OUTPUT_FILE.read_text(encoding="utf-8")
    print(f"Successfully read from {OUTPUT_FILE}:")
    print(f"-> {content}")
else:
    print(f"Error: {OUTPUT_FILE} does not exist.")
```
