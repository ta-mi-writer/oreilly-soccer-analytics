# Kaggle ノートブック環境プローブ（ディスク / RAM / Kaggle API 認証）

MOSS-Image-SFT（27.2 GB）をノートブック内から Kaggle Models/Datasets にアップロードできるかを調べるためのプローブです。

```python
import shutil
import os

for path in ["/kaggle/working", "/kaggle/input", "/"]:
    try:
        usage = shutil.disk_usage(path)
        print(f"{path}: total={usage.total/1e9:.1f}GB free={usage.free/1e9:.1f}GB")
    except Exception as e:
        print(f"{path}: ERROR {e}")

with open("/proc/meminfo") as f:
    for line in f:
        if line.startswith(("MemTotal", "MemAvailable")):
            print(line.strip())
```

```python
import os

for key in ["KAGGLE_USERNAME", "KAGGLE_KEY", "KAGGLE_API_TOKEN", "KAGGLE_KERNEL_RUN_TYPE"]:
    print(f"{key}: {'SET' if os.environ.get(key) else 'not set'}")
```

```python
import subprocess

res = subprocess.run(["kaggle", "datasets", "list", "-s", "moss"], capture_output=True, text=True, timeout=120)
print("returncode:", res.returncode)
print(res.stdout[:2000])
print(res.stderr[:2000])
```
